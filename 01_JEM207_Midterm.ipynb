```python
#excersize 1
import pandas as pd

acn = pd.read_csv("ACN.csv", index_col="Date", parse_dates=["Date"], date_format="%y-%m-%d")
acn["Ticker"] = "ACN"

all_col = pd.read_csv("ALL.csv", index_col="Date", parse_dates=["Date"], date_format="%y-%m-%d")
all_col["Ticker"] = "ALL"

ko = pd.read_csv("KO.csv", index_col="Date", parse_dates=["Date"], date_format="%y-%m-%d")
ko["Ticker"] = "KO"

expe = pd.read_csv("EXPE.csv", index_col="Date", parse_dates=["Date"], date_format="%y-%m-%d")
expe["Ticker"] = "EXPE"

#Consolidated
prices = pd.concat([acn, all_col, ko, expe])
prices.index = pd.to_datetime(prices.index)

#part a
no_diff = prices[prices["High"] - prices["Low"] == 0] # should be the difference 
print(no_diff.size) # since this is zero then every stock per entity had a price that was had a high spread
```

```python
#excersize 2
for ticker, df in prices.groupby("Ticker"):
    high = df["High"].max()
    low = df["Low"].min()
    print(f"{ticker} had {high} as its highest price and {low} as its lowest")
```

```python
#excersize 3 
import numpy as np

#calculate the logarithm of the close
log_prices = np.log(prices["Close"])
log_returns = np.diff(log_prices, axis=0)

prices.groupby("Ticker")["Close"].agg(["min", "max", "mean", "std"])
```

```python
for ticker, df in prices.groupby("Ticker"):
    print(df['High'].nlargest(2).idxmax(0))
    print(df['Low'].nlargest(2).idxmin(0))
```

```python
#excersize 5
weekly = prices.groupby("Ticker").resample("W")["Volume"].sum()
median_weekly_volume = weekly.groupby(level=0).median()
print(median_weekly_volume)
```
